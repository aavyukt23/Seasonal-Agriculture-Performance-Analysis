## Data Cleaning and Preparation

Check missing values

```python
df.isnull().sum()
```

Dataset has missing values in:

- Rainfall
- Soil Moisture
- Yield

Handle missing values

```python
df["Rainfall_mm"] = df["Rainfall_mm"].fillna(
    df["Rainfall_mm"].median()
)

df["Soil_Moisture_pct"] = df["Soil_Moisture_pct"].fillna(
    df["Soil_Moisture_pct"].median()
)

df["Yield_Tonnes_Ha"] = df["Yield_Tonnes_Ha"].fillna(
    df["Yield_Tonnes_Ha"].median()
)
```

```python
df.duplicated().sum()
```

## Identify Significant Differences

ANOVA

```python
from scipy.stats import f_oneway

kharif = df[df["Season"] == "Kharif"]["Yield_Tonnes_Ha"]
rabi = df[df["Season"] == "Rabi"]["Yield_Tonnes_Ha"]
zaid = df[df["Season"] == "Zaid"]["Yield_Tonnes_Ha"]

result = f_oneway(kharif, rabi, zaid)

print("F-statistic:", result.statistic)
print("p-value:", result.pvalue)
```

## Key Findings

Based on the analysis of seasonal agricultural performance, the following key findings were identified:

1. **Seasonal variation in yield:**
   Kharif recorded the highest average yield at **5.63 tonnes/ha**, followed by Rabi at **5.04 tonnes/ha** and Zaid at **4.64 tonnes/ha**. The median yield followed the same pattern, with Kharif at **1.94 tonnes/ha**, Rabi at **1.67 tonnes/ha**, and Zaid at **1.46 tonnes/ha**.

2. **Environmental conditions vary considerably across seasons:**
   Kharif had the highest average rainfall (**849.20 mm**) and soil moisture (**31.15%**), while Zaid had the highest average temperature (**31.04°C**) and sunlight duration (**8.18 hours/day**). Rabi showed intermediate environmental conditions for most variables.

3. **Resource usage is relatively consistent across seasons:**
   Average nitrogen, phosphorus, potassium, fertilizer, and pesticide usage were relatively similar across all three seasons. However, water usage varied, with Zaid having the highest average water usage (**6,419.89 m³**) and Rabi the lowest (**5,846.99 m³**).

4. **Economic performance differs strongly across seasons:**
   Kharif recorded the highest average revenue (**₹710,719.06**) and average profit (**₹178,914.65**). Rabi had an average profit of **₹87,689.47**, while Zaid recorded a negative average profit of **₹24,804.82**.

5. **Seasonal differences in costs:**
   Zaid had the highest average total cost (**₹543,976.73**), while Rabi had the lowest (**₹513,836.58**). The combination of higher costs and lower revenue contributed to the negative average profit observed in Zaid.

6. **Market prices were relatively similar:**
   Average market prices were relatively close across the three seasons, ranging from approximately **₹44,212 to ₹44,851 per tonne**. Therefore, the differences in economic performance appear to be associated with factors beyond market price alone, such as production, costs, and seasonal conditions.

7. **Statistical analysis of yield:**
   The one-way ANOVA produced an **F-statistic of 1.544** and a **p-value of 0.214**. Since the p-value is greater than 0.05, there is **insufficient statistical evidence to conclude that the mean yield differs significantly across Kharif, Rabi, and Zaid seasons**.


## Recommendations

Based on the findings of the analysis, the following recommendations can be made:

1. **Adopt season-specific agricultural planning:**
   Agricultural planning should consider the differences in rainfall, temperature, sunlight, and soil moisture observed across seasons.

2. **Focus on efficient water management:**
   Zaid recorded the highest average water usage while also having the lowest average yield. Therefore, water usage during Zaid should be monitored carefully and managed efficiently.

3. **Improve economic planning for Zaid:**

4. **Use crop-season comparisons for crop selection:**
   Since different crops may perform differently across seasons, crop selection should consider the season in which each crop demonstrates better performance.

5. **Consider regional differences:**
   The observed differences in yield across states suggest that seasonal agricultural planning should also consider geographical and regional conditions.

6. **Monitor environmental conditions:**
   Rainfall, temperature, humidity, sunlight, and soil moisture should be monitored regularly to support better seasonal agricultural decisions.

7. **Use evidence rather than season alone:**
   Although Kharif had the highest observed average yield, the ANOVA result did not show a statistically significant difference in mean yield across seasons. Therefore, agricultural decisions should consider other factors such as crop type, environmental conditions, resource usage, and economic outcomes rather than relying only on season.


## Conclusion

The analysis of the Seasonal Agriculture Performance dataset demonstrates that agricultural conditions and outcomes vary across Kharif, Rabi, and Zaid seasons.

Kharif showed the highest observed average yield, revenue, and profit, while Zaid showed the lowest average yield and a negative average profit. Environmental conditions also varied considerably, particularly in rainfall, temperature, sunlight, and soil moisture. Resource usage was relatively consistent for fertilizers and nutrients, although water usage was highest during Zaid.

The crop-season and state-season comparisons provide further insight into differences in agricultural performance across crops and geographical regions. The correlation analysis also helps identify relationships between environmental conditions, resource usage, yield, and economic outcomes.

However, the ANOVA result produced a p-value of **0.214**, which is greater than 0.05. Therefore, the analysis does not provide sufficient statistical evidence that mean agricultural yield differs significantly across the three seasons.

Overall, the findings show that seasonal conditions are important for agricultural planning, but effective decision-making should also consider crop characteristics, regional conditions, resource requirements, production costs, and economic returns.
